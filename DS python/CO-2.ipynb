```python
import matplotlib.pyplot as plt
y1=[]#stores positive values
y2=[]#stores negative values
x= range(-100,100,10)
for i in x: y1.append(i**2)
for i in x: y2.append(-i**2)
plt.plot(x,y1)
plt.plot(x,y2)
plt.xlabel("X")
plt.ylabel("Y")
plt.ylim(-2000,2000)#limits of y-axis
plt.axhline(0)
plt.axvline(0)
plt.savefig("quad.png")
plt.show()
```

```python
import matplotlib.pyplot as plt
ages=[2,5,70,40,30,45,50,45,43,40,44,60,57,70,80,90,100,22,24,17,18,19]
range=(0,100)
bins=10
plt.hist(ages,bins,range,color="blue ",histtype="bar",rwidth=0.8)
plt.xlabel("Age")
plt.ylabel("No of people")
plt.title("Histogram")
plt.show()
```
